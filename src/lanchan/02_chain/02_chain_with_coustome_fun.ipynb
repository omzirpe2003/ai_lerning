```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

promot_templet = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistent"),
    ("human", "{input}")
])



## Task 2

llm_openai=ChatOpenAI(model='gpt-5-mini',temperature=0)


## Task: 3  Str output parser .content
str_parser = StrOutputParser()



from langchain_core.runnables import RunnableLambda

## Task 4:- create a dic maker
def dic_maker(text:str)-> dict:
    return {"text": text}

## Conver that fun to runnable lambda
dict_maker_runnable = RunnableLambda(dic_maker)


## Task 5:- Create a Templet for post

promot_post =ChatPromptTemplate.from_messages(
    messages =[
        ("system","You are a social media post genrater"),
        ("human","create a post for the follwing {text} ")
    ]
)



## Task 6 : LLM

llm_openai= ChatOpenAI(model='gpt-5-mini')


## Task 7 : Str parser

str_parser= StrOutputParser()


## Task 8 : Create Chain

chain = promot_templet | llm_openai | str_parser |  dict_maker_runnable | promot_post | llm_openai | str_parser

chain.invoke("What is the capital of france")
```
