```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

promot_templet = ChatPromptTemplate.from_messages([
    ("system","You movie summerizer"),
    ("human", "Plz summerize the movie in brief {input}")
])



## Task 2

llm_openai=ChatOpenAI(model='gpt-5-mini',temperature=0)


## Task: 3  Str output parser .content
str_parser = StrOutputParser()



from langchain_core.runnables import RunnableLambda

## Task 4:- create a dic maker
def dic_maker(text:str)-> dict:
    return {"text": text}

## Conver that fun to runnable lambda
dict_maker_runnable = RunnableLambda(dic_maker)




```

## Paraller chain 1 ##

```python

## Task 1 ##

linkdin_promot = ChatPromptTemplate.from_messages([
    ("system","your are a Linkdin post genrater"),
    ('human',"Create a post for follwing text for linkedin : {text}")
])


llm_openai=ChatOpenAI(model='gpt-5-mini',temperature=0)

str_parser= StrOutputParser()

chian_linkedin = linkdin_promot | llm_openai | str_parser
```

```python
## Paraller chain 2 ##
```

```python
def insta_chain(text:str):

    text=text['text']
    insta_promot = ChatPromptTemplate.from_messages([
        ("system","your are a instagram post genrater"),
        ('human',"Create a post for follwing text for instagram : {text}")
    ])

    llm_openai=ChatOpenAI(model='gpt-5-mini',temperature=0)

    str_parser= StrOutputParser()   

    chain_insta = insta_promot | llm_openai | str_parser

    result =chain_insta.invoke(text)

    return result


insta_chain_runnable = RunnableLambda(insta_chain)
```

## Final Orchastration ##

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel

final_chain = (promot_templet | llm_openai | str_parser | dict_maker_runnable | RunnableParallel(branches = {'linkedin': chian_linkedin, 'insta':insta_chain_runnable}))

result = final_chain.invoke("KGF")
result['branches']['linkedin']
```

## Chain as a runnable ##

```python
## TASK 1 ## 

def beufity (final_response : dict)-> dict:
    linkedin_response = final_response['branches']['linkedin']
    insta_response = final_response['branches']['insta']
    return {'insta':insta_response,'linkedin':linkedin_response}

final_result = RunnableLambda(beufity)


beutify_chain = final_chain | final_result

beutify_chain.invoke("ABCD-2")
```
