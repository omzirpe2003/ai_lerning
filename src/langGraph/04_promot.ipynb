## Promots in LangGraph

```python

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
import os
load_dotenv()

if os.environ['OPENAI_API_KEY']:
    print("Openai key is set")
else:
    raise ValueError("openAi key is not set")


llm_openai = ChatOpenAI(model='gpt-5-mini',temperature=0)

```

```python

from langchain_core.prompts import ChatPromptTemplate

my_promot = ChatPromptTemplate.from_messages(
    [
        ('system','You are a helpful assistant that helps to create and curate posts for a social media platform.'),
        ('user','want to create post on {topic}')
    ]
)


chain = my_promot | llm_openai

chain.invoke({'topic':"The importance of Gaming privacy in the digital age"})
```
