```python
from langchain_openai import ChatOpenAI

llm_openai=ChatOpenAI(model="gpt-5-mini",temperature=0)
```

## First Chain ##

```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

promot_templet = ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistent"),
    ("human", "{input}")
])



## Task 2
llm_openai=ChatOpenAI(model='gpt-5-mini',temperature=0)


## Task: 3  Str output parser .content
str_parser = StrOutputParser()

```

## Manual Invocation 

```python

templet = promot_templet.invoke({"input":"What is the capital of France?"}) 

result = llm_openai.invoke(templet)

final_output = result.content
```

## Chain Invocation

```python
chain = promot_templet | llm_openai | str_parser 

chain.invoke({"input":"What is capital of india"})
```

## Using RunnableSequence ## 

```python
from langchain_core.runnables import RunnableSequence

chain= RunnableSequence(promot_templet,llm_openai,str_parser)

chain.invoke({"input":"What is capital of india"})
```
